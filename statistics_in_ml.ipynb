```python
col=df["total_bill"]

mean_val = np.mean(col)
meadian_val = np.median(col)
mode_val = col.mode()[0]
print(f"mean value is {mean_val:.2f}")
print(f"median value is {meadian_val:.2f}")
print(f"mode value is {mode_val:.2f}")

plt.figure(figsize=(5,3))
sns.histplot(col, kde=True, color="skyblue",)
plt.axvline(mean_val,color='red', linestyle="--", label=f"Mean:{mean_val:.2f}")
plt.axvline(meadian_val,color='green', linestyle="--", label=f"Median:{meadian_val:.2f}")
plt.axvline(mode_val,color='yellow', linestyle="--", label=f"Mode:{mode_val:.2f}")
plt.legend()
plt.title("Distribution of total bill")
plt.show()

```

```python
col1=df["tip"]

mean_val = np.mean(col1)
meadian_val = np.median(col1)
mode_val = col1.mode()[0]
print(f"mean value is {mean_val:.2f}")
print(f"median value is {meadian_val:.2f}")
print(f"mode value is {mode_val:.2f}")

plt.figure(figsize=(5,3))
sns.histplot(col1, kde=True, color="tan",)
plt.axvline(mean_val,color='red', linestyle="--", label=f"Mean:{mean_val:.2f}")
plt.axvline(meadian_val,color='green', linestyle="--", label=f"Median:{meadian_val:.2f}")
plt.axvline(mode_val,color='yellow', linestyle="--", label=f"Mode:{mode_val:.2f}")
plt.legend()
plt.title("Distribution of tip")
plt.show()

```
